```python
#3.1
import random as r
r.seed(1024170282)
l=[]
for i in range(0,100):
    l.append(r.randint(100,900))
print("random list",l)

#3.2
count=0
odd=[]
for j in range(0,100):
    if l[j]%2!=0:
        count+=1
        odd.append(l[j])
print("count of odd no",count)
print("odd numbers list",odd)

#3.3
even=0
evenn=[]
for j in range(0,100):
    if l[j]%2==0:
        even+=1
        evenn.append(l[j])
print("count of even no",even)
print("even numbers list",evenn)

#3.4
countprime=0
primee=[]
for i in range(0,100):
    prime=True
    if l[i] < 2:
        prime = False
    else:
        for j in range (2,l[i]//2+1):
            if l[i]%j==0:
                prime=False
                break
    if prime:
        primee.append(l[i])
        countprime+=1
print("no of prime no",countprime)
print("prime numbers list",primee)


#3.5
mostfreq=l[0]
for i in l:
    if l.count(i)>l.count(mostfreq):
        mostfreq=i
print("most frequently occ element",mostfreq, "no of times: ",l.count(mostfreq))
        
   
```

```python
#4.1
digits=[1,0,2,4,1,7,0,2,8,2]
A={digit*7 for digit in digits}
B={digit*9 for digit in digits}

print("set A",A)
print("set B",B)

#4.2
print("union of a and b is", A.union(B))

#4.3
print("intersection of a and b is", A.intersection(B))

#4.4
# Symmetric diff means elements that are in either A or B, but NOT in both.
# a-b diff means elements in A but NOT in B.
print("A-B:", A.difference(B))
print("B-A:", B.difference(A))

#4.5
print("Symmetric difference:", A.symmetric_difference(B))

#4.6
print("Is A a subset of B?", A.issubset(B))
print("Is B a superset of A?", B.issuperset(A))

#4.7
x=int(input("Enter a value: "))
A.discard(x)
print("Set A after discard:", A)
```

```python
my_dict={
    "name":"Bhoomi Mahna",
    "roll no":1024170282,
    "branch":"COPC",
    "age":20,
    "city":"Patiala"
}
#5.1
my_dict["location"]=my_dict.pop("city")
print("After renaming city:",my_dict)

#5.2
my_dict["cgpa"]=9.7
print("After adding CGPA:", my_dict)

#5.3
my_dict["age"]=my_dict["age"]+1
print("After increasing age",my_dict)

#5.4
d1=my_dict.copy()
removed=d1.pop("branch")
print("Using pop ", d1)
print("Removed value ", removed)

d2=my_dict.copy()
del d2["branch"]
print("Using del ", d2)
# pop() returns the removed value, whereas del only deletes the key-value pair.

#5.5
for key,val in my_dict.items():
    print(key,val)

#5.6
if "email" in my_dict:
    print("email ", my_dict["email"])
else:
    print("email key does not exist")

#5.7
# When the same key exists, the value from the dictionary written LAST wins.
friend_dict = {
    "name": "Kishna Gautam",
    "roll_no": "1024070300",
    "branch": "COE",
    "age": 21,
    "city": "UK"
}
mergeddic={**my_dict,**friend_dict}
print("merged dictionary", mergeddic)


#5.8
newd={
    key:value
    for key,value in my_dict.items()
    if isinstance(value,str)
}
print("Only string values", newd)
```
