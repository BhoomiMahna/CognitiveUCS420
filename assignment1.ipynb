```python
n = int(input("Enter number"))
sum = 0
for num in range(2,n+1):
    prime=True
    for i in range(2,num//2+1):
        if num%i==0:
            prime=False
            break

    if prime:
        sum+=num

print(sum)
```

```python
n=int(input("enter number"))
def oddsum(n):
    sum=0
    for i in range(1,n+1):
        if(i%2!=0):
            sum+=i
    return sum

print(oddsum(n))
```

```python
n = int(input("Enter number"))
def primesum(n):
    sum = 0
    for num in range(2,n+1):
        prime=True
        for i in range(2,num//2+1):
            if num%i==0:
                prime=False
                break

        if prime:
            sum+=num
    return sum
print(primesum(n))
```
