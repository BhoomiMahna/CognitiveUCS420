```python
#q1
import numpy as np
temperature = np.array([25, 28, 31, 35, 38, 27, 33, 40])

#1.1
temp=temperature+2
print(temp)
#1.2
tempF=temperature*(9/5)+32
print(tempF)
#1.3
temp32=temperature[temperature>32]
print(temp32)
#1.4
summ=np.sum(temperature>32)
print(summ)
#1.5
#vectorization- Instead of using a for loop to add 2 to each temperature, numpy performs the operation on the entire array at once
#boolean- Instead of checking each temperature individually, numpy creates a boolean condition and directly selects the values satisfying it.
```

```python
#q2
steps = np.array([
[5000, 6200, 7100],
[8000, 7500, 9000],
[4500, 5100, 4800],
[9000, 8500, 9500]])

#2.1
print("total steps are ",np.sum(steps))
#2.2
print("mean number of steps are ",np.mean(steps))
#2.3
print("max and min values of steps are ",np.max(steps)," and ",np.min(steps), " respectively")
#2.4
print("steps each day ",np.sum(steps,axis=0))
#2.5
print("total steps of each user ", np.sum(steps,axis=1))
#2.6
total_each_user=np.sum(steps, axis=1)
max_pos=np.argmax(total_each_user)
print("user with max steps ",max_pos, "and user number is ", max_pos+1)
      
```

```python
#q4
X = np.array([
    [6, 70, 3],
    [5, 50, 6],
    [8, 80, 2],
    [4, 30, 8]
])
y = np.array([40, 65, 30, 85])


#4.1
print("Shape of X ", X.shape)
print("Dimensions of X ", X.ndim)

#4.2
XT = X.T
print(XT)
#Transpose changes rows into columns and columns into rows. X has shape (4, 3), while X.T has shape (3, 4).

#4.3
XTX = XT @ X
print(XTX)

#4.4
XTX_inverse = np.linalg.inv(XTX)
print(XTX_inverse)

#4.5
beta = XTX_inverse @ XT @ y
print(beta)

#4.6
#Interpretation of coefficients:
#The first coefficient represents the effect of sleep hours on the assistance score, keeping the other features constant.
#The second coefficient represents the effect of activity level on the assistance score, keeping the other features constant.
#The third coefficient represents the effect of stress level on the assistance score, keeping the other features constant.

#4.7
new_user = np.array([5, 40, 7])
predicted_score = new_user @ beta
print("New user ", new_user)
print("Predicted assistance score ", predicted_score)
```
