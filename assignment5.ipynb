```python
#q1
import numpy as np
a=np.array([10,20,30,40,50])
#1.1
a=a+2
print(a)
#1.2
a=a*3
print(a)
#1.3
a=a/2
print(a)
```

```python
#q2
#2.1
arr=np.array([1,2,3,4,5])
arr=np.flip(arr)
print("reversed array is ",arr)

#2.2
x=np.array([1,2,3,4,5,1,2,1,1,1])
y=np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3, ]) 

value1,count1=np.unique(x,return_counts=True)
mse1=value1[np.argmax(count1)]
idx1=np.where(x==mse1)[0]

value2,count2=np.unique(y,return_counts=True)
mse2=value2[np.argmax(count2)]
idx2=np.where(y==mse2)[0]

print("most frequent element in x is ",mse1," and its index is ",idx1)
print("most frequent element in y is ",mse2," and its index is ",idx2)
```

```python
#q4
bhoomi=np.linspace(10,100,25)
print(bhoomi.ndim)
print(bhoomi.shape)
print(bhoomi.size)
print(bhoomi.dtype)
print(bhoomi.nbytes)

transpose_arr=bhoomi.reshape(1,25).T
print(transpose_arr)
print(bhoomi.T)
#For a 1-D array,.T does not convert it into a column/row matrix.
```
